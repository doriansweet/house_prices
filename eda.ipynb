# House Prices - Advanced Regression Techniques

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
pd.set_option('display.max_columns', None)
```

```python
home_data = pd.read_csv('house-prices-advanced-regression-techniques/train.csv')
home_data.head()
```

```python
home_data.info()
```

```python
print(f'Количество домов: {home_data.shape[0]}')
print(f'Количество признаков: {home_data.shape[1]}')
print(f'Дубликаты: {home_data.duplicated().sum()}')
```

```python
home_data = home_data.drop('Id', axis=1)
```

```python
home_data.describe()
```

Заполняю нуллы

```python
home_null = home_data.isna().mean() * 100
home_null = home_null[home_null>0].sort_values(ascending=False)
home_null
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    x=home_null.values,
    y=home_null.index
)

plt.xlabel('Пропущено, %')
plt.ylabel('')
plt.title('Доля пропусков до обработки')
plt.tight_layout()
plt.show()
```

```python
home_data.loc[
    home_data['MasVnrArea'].isna(),
    ['MasVnrType', 'MasVnrArea']
]
```

MSSubClass и MoSold представлены числами, но являются категориальными признаками. Преобразуем их в строки, чтобы get_dummies() создал для их значений отдельные dummy-переменные.

```python
home_data['MSSubClass'] = (
    home_data['MSSubClass']
    .astype(str)
)

home_data['MoSold'] = (
    home_data['MoSold']
    .astype(str)
)
```

```python
cols = ['PoolQC', 'MiscFeature', 'Alley', 'Fence', 'MasVnrType', 'FireplaceQu', 'GarageType', 'GarageFinish', 'GarageQual', 'GarageCond', 'BsmtFinType2', 'BsmtQual', 'BsmtCond', 'BsmtFinType1']
home_data[cols] = home_data[cols].fillna('No')
cols_numeric = ['MasVnrArea', 'GarageYrBlt']
home_data[cols_numeric] = home_data[cols_numeric].fillna(0)
```

```python
home_data['BsmtExposure'] = home_data['BsmtExposure'].fillna('No Basement')
```

```python
home_data['Electrical'] = home_data['Electrical'].fillna(home_data['Electrical'].mode()[0])
```

```python
home_data['LotFrontage'] = (home_data['LotFrontage']
                               .fillna(home_data
                                      .groupby('Neighborhood')['LotFrontage']
                                      .transform('median')))
```

```python
remaining_nulls = (
    home_data
    .isna()
    .sum()
    .sort_values(ascending=False)
)

remaining_nulls[remaining_nulls > 0]
```

```python
home_data_dummies = pd.get_dummies(home_data, drop_first=True)
home_data_dummies.head()
```

```python
correlation = (
    home_data_dummies
    .corr(numeric_only=True)['SalePrice']
    .drop('SalePrice')
)

top_features = correlation.abs().nlargest(20).index
top_correlations = correlation.loc[top_features].sort_values()

colors = [
    'crimson' if value < 0 else 'steelblue'
    for value in top_correlations
]

plt.figure(figsize=(10, 7))

plt.barh(
    top_correlations.index,
    top_correlations.values,
    color=colors
)

plt.axvline(0, color='black', linewidth=1)
plt.xlabel('Корреляция с SalePrice')
plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots( figsize=(12, 4))

sns.histplot(
    home_data['SalePrice'],
    kde=True,
)
ax.set_title('Распределение SalePrice')

plt.tight_layout()
plt.show()
```

```python
features = [
    'OverallQual',
    'GrLivArea',
    'GarageCars',
    'GarageArea',
    'TotalBsmtSF',
    '1stFlrSF',
    'YearBuilt'
]

fig, axes = plt.subplots(3, 3, figsize=(15, 12))
axes = axes.flatten()

for ax, feature in zip(axes, features):
    sns.scatterplot(
        data=home_data,
        x=feature,
        y='SalePrice',
        alpha=0.6,
        ax=ax
    )
    ax.set_title(feature)

for ax in axes[len(features):]:
    ax.remove()

plt.tight_layout()
plt.show()
```

```python
neighborhood_order = (
    home_data
    .groupby('Neighborhood')['SalePrice']
    .median()
    .sort_values()
    .index
)

plt.figure(figsize=(12, 8))

sns.boxplot(
    data=home_data,
    x='SalePrice',
    y='Neighborhood',
    order=neighborhood_order
)

plt.title('Стоимость домов по районам')
plt.tight_layout()
plt.show()
```

## Выводы EDA

1. Целевая переменная SalePrice имеет выраженную правостороннюю
   асимметрию, поэтому для обучения будет использован log1p(SalePrice).

2. Пропуски в PoolQC, Alley, Fence, FireplaceQu и гаражных признаках
   преимущественно означают отсутствие соответствующего объекта,
   а не случайно потерянные данные.

3. Наиболее сильная положительная связь с ценой наблюдается у
   OverallQual, GrLivArea, GarageCars и TotalBsmtSF.

4. В GrLivArea присутствуют потенциальные выбросы: дома с очень
   большой площадью и относительно низкой стоимостью.

5. Категориальный признак Neighborhood заметно связан с ценой дома.
